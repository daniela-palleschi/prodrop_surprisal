# Colab + GitHub setup

The open-source LLM we want to use is too large to run on my current machine. I've opted to run it on Google Colab, which does not store files between sessions. My solution is to use a GitHub repo to pull and push changes from Colab.

Run this once at the start of every Colab session to clone the repo, and again whenever you want to push changes back to GitHub. Keep this separate from your working notebooks so you don't need to redo it each time you edit a script. To use this script you need:

- a Colab account (colab.research.google.com)
- a GitHub account and relevant repo
- a Personal Access Token (PAT) - Classic with `repo` checked
- Colab 'secrets' for: `email`, `name`, `github_token` (where value = your PAT)

## Clone repo

Pulls the full repo into the Colab filesystem so all files are available. Do this at the start of every fresh session (a runtime restart wipes the filesystem). If you haven't cloned this repo into Colab before, open a new notebook, paste the clone cell below into it, and run. If you have, then when you open Colab you should see this file (`colab_setup.ipynb`) under 'Recents'. Just click it and run.

> [!NOTE]
> In Colab, code preceded with `!` is run as bash/shell, not Python.

```python
REPO_URL = "https://github.com/daniela-palleschi/prodrop_surprisal.git"
REPO_NAME = "prodrop_surprisal"

!git clone {REPO_URL}
%cd {REPO_NAME}
```

```python
# sanity check: confirm you're in the right place before doing anything else
!pwd
!git remote -v
```

## Push changes

Run this after editing any file you want saved back to GitHub. First set up two Colab Secrets (key icon, left panel):

- `email`: your real email (used only locally for the git commit, never written to the repo)
- `name`: your real name
- `github_token`: a GitHub Personal Access Token with `repo` scope (GitHub > Settings > Developer settings > Personal access tokens > Tokens classic > Generate new token)

Turn on Notebook access for each secret. This avoids putting your email or a token directly in the notebook.

```python
from google.colab import userdata

token = userdata.get('github_token')
email = userdata.get('email')
name = userdata.get('name')

!git config --global user.email "{email}"
!git config --global user.name "{name}"

COMMIT_MESSAGE = "update from Colab"  # edit this each time

!git add -A
!git commit -m "{COMMIT_MESSAGE}"
!git push https://{token}@github.com/daniela-palleschi/prodrop_surprisal.git
```
